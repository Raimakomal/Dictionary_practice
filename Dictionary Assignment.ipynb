# Implement nested to nested to nested dictionary:

```python
d={'Electronics': {'Laptop': {'Brand1': 'Dell', 'Brand2': 'HP', 'Brand3': 'ASER', 'Brand4': 'TOSHIBA'},
                       'Mobile': {'C1': 'NOKIA', 'C2':'INFINIX', 'C3': 'SAMSUNG', 'C4':'OPPO'}},
       'Makeup':{'Foundation':{'F1':'MISSROSE','F2':'HUDA BEAUTY', 'F3':'LOREAL','F4':'MAC'},
                 'Lipstick':{'L1':'MEDORA', 'L2':'HUDA BEAUTY', 'L3':'LOREAL', 'L4':'FLARMAR'}}}                                
print(d)
```

# Print all mobile types

```python
for key, value in d['Electronics']['Mobile'].items():  # used items method
    print(key,value)
```

# Print all laptop types

```python
for k,v in d['Electronics']['Laptop'].items():
    print(k,v)
```

# Print all makeup products

```python
for k,v in d['Makeup'].items():
    print(k,v)
```

# Print only mac from dictionary

```python
print(d['Makeup']['Foundation']['F4'])
```

# Add new laptop mobile and makeup type

# Remove MEDORA from dictionary.

# Print only Brand3, C4, F4, L2

```python
print('Brand3:', d['Electronics']['Laptop']['Brand3'])
print('C4:', d['Electronics']["Mobile"]['C4'])
print('F4:', d['Makeup']['Foundation']['F4'])
print('L2:', d['Makeup']['Lipstick']['L2'])
```

# Add new complete category and sub category

```python
d['Food'] = {
    "Fruits":{'FR1': 'APPLE','FR2': 'BANANA','FR3':'FIG', 'FR4':'GRAPES'},
    'Vegetables':{'V1': 'CARROT', 'V2':'ONION', 'V3':'POTATO', 'V4':'CABBAGE'}
}
print(d)
```

# Remove any laptop, mobile, foundation and lipstick

```python
d["Electronics"]['Laptop'].popitem()
d["Electronics"]['Mobile'].popitem()
d["Makeup"]['Foundation'].popitem()
d["Makeup"]['Lipstick'].popitem()
```

# Print those products which start from ‘M’

```python
for cat,sub in d.items():
    for s, items in sub.items():
        for k, v in items.items():
            if v.startswith('M'):
                print(v)
```

# Print those products which contained ‘M’ and ‘E’

```python
for cat,sub in d.items():
    for s,items in sub.items():
        for k,v in items.items():
            if 'M' in v and 'E' in v:
                print(v)
```

# Print products which contained No. of characters less than 4

```python
for cat,sub in d.items():
    for sub,items in sub.items():
        for k,v in items.items():
            if len(v)<4:
                print(v)
```

# Print product with number of characters

```python
for cat,sub in d.items():
    for sub,items in sub.items():
        for k,v in items.items():
            print(v,"=",len(v))
```

# Print those Mobile names which contained ‘I’

```python
mobiles=d.get('Electronics',{}).get('Mobile',{})
for n in mobiles.values():
    if 'I' in n:
        print(n)
```

# Print foundations which contained ‘Y’ at last position

```python
foundations=d.get("Makeup",{}).get("Foundation",{})
for n in foundations.values():
    if n.endswith("Y"):
        print(n)
```
